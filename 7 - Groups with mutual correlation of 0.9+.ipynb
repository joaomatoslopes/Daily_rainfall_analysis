```python
# Importing libraries
import pandas as pd
%matplotlib inline
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns; sns.set()
from sklearn.datasets import load_iris
from sklearn.preprocessing import StandardScaler 
import warnings
warnings.filterwarnings("ignore")
import scipy
from PIL import Image
import statsmodels.api
from itertools import combinations
import time
import datetime
import pickle
import os
```

```python
# Importing coordinates
local = pd.read_csv(r"C:\Users\joaoa\Desktop\Doutoramento\Professora\Nova pasta\Todos os Pontos (temperatura)\local.cvs")

# Defining parameters
min_corr = 0.9 ##########
Y = range(1950, 2023) # 1950-2023
K = range(2, 7) # 2-7
T = ["no restriction", "9h-9h"]
H = [1,2,4,6,12,24,48,96]
```

```python
max_precipitations = pd.read_csv(r"D:\Precipitation\by_water_year\max_precipitations_(mm).csv")
max_precipitations
```

```python
# Collecting all information in one dictionary
dict_max = {"no restriction" : {}, "9h-9h" : {}}

for t in T: 
    for hours in H:
        dict_max[t][str(hours) + "h"] = max_precipitations[(max_precipitations["type_of_period"] == t) & (max_precipitations["duration"] == str(hours) + "h")].iloc[:, 3:].reset_index(drop = True)
```

```python
dict_max["no restriction"]["2h"]
```

```python
dict_corr = {"no restriction" : {}, "9h-9h" : {}}

for t in T:
    for h in H:
        df = dict_max[t][str(h) + "h"]

        # Standardizing dataframe
        scaler = StandardScaler() 
        variables = scaler.fit_transform(df) 
        variables = pd.DataFrame(variables, columns = df.columns)
        variables = variables*np.sqrt((len(variables)-1)/len(variables))
        var_list = variables.columns

        reset_columns = []
        for _ in range(0,1012):
            reset_columns.append(str(_))
        variables.columns = reset_columns
        dict_corr[t][str(h) + "h"] = variables.corr()
```

```python
# Checking if there is already saved information
if os.path.isfile(r"C:\Users\joaoa\Desktop\docs\dict_resume_9.pkl"): ##########
    
    # Uploading dictionary
    with open(r"C:\Users\joaoa\Desktop\docs\dict_resume_9.pkl", "rb") as h: ##########
        dict_resume = pickle.load(h)
        
    # Reporting resume
    for t in T:
        for n in dict_resume["clusters"][t].keys():
            print("Lenght of dictionary " + str(n) + " (" + t + "):", len(dict_resume["clusters"][t][n]))
        print("Number of rows that are still being considered (" + t + "):", len(dict_resume["rows"][t]))
        
        
else:  
    dict_resume = {"clusters" : {"no restriction" : {}, "9h-9h" : {}}, 
                   "rows" : {"no restriction" : [], "9h-9h" : []}}
    
    # dict_resume["clusters"][n=1]
    for t in T:

        dict_resume["clusters"][t][1] = set()
        for _ in range(0,1012):

            dict_resume["clusters"][t][1].add(tuple([_]))
            dict_resume["rows"][t].append(_)
        print("Lenght of dictionary 1 (" + t + "):", len(dict_resume["clusters"][t][1]), "| Current time:", datetime.datetime.now().time(), "\n")

    # Saving dictionary
    with open(r"C:\Users\joaoa\Desktop\docs\dict_resume_9.pkl", "wb") as h: ##########
        pickle.dump(dict_resume, h)
```

```python
time_consumed = []

# dict_clusters[n>1]
for t in T:

    n = len(dict_resume["clusters"][t])+1
    while len(dict_resume["clusters"][t][n-1]) != 0:
        
        i = time.time()
        i_percent = time.time()
        time_consumed_percent = []
        dict_resume["clusters"][t][n] = set()
        p = 0
        
        for_row_list = dict_resume["rows"][t].copy()
        
        r = 0
        for row in for_row_list:
            
            len_i = len(dict_resume["clusters"][t][n])

            if (100*r) // len(for_row_list) > p:
                p = (100*r) // len(for_row_list)
                time_consumed_percent.append(time.time()-i_percent)
                i_percent = time.time()
                print("Completed", p, "% | Lenght:", len(dict_resume["clusters"][t][n]), "| Current time:", datetime.datetime.now().time(), "| Duration:", round(time_consumed_percent[-1]/60,1), "min")                
            r += 1
                
            for tup in dict_resume["clusters"][t][n-1]:

                if row in tup:
                    continue

                check = 0
                for col in tup:
                    for h in H:
                        if dict_corr[t][str(h) + "h"].iloc[row,col] < min_corr:
                            check = 1
                            break
                    if check == 1:
                        break

                if check == 0:
                    lista = list(tup + (row,))
                    lista.sort()
                    dict_resume["clusters"][t][n].add(tuple(lista))

            if len(dict_resume["clusters"][t][n]) == len_i:
                dict_resume["rows"][t].remove(row)
                
        # Saving dictionary
        with open(r"C:\Users\joaoa\Desktop\docs\dict_resume_9.pkl", "wb") as h: ##########
            pickle.dump(dict_resume, h)
            
        # Time prediction
        time_consumed.append((time.time()-i)/len(dict_resume["clusters"][t][n-1]))
        print("Lenght of dictionary " + str(n) + " (" + t + "):", len(dict_resume["clusters"][t][n]), "| Current time:", datetime.datetime.now().time(), "| Expected next result in", round(np.mean(time_consumed)*len(dict_resume["clusters"][t][n])/60,1), "min", "\n")

        n += 1
```
