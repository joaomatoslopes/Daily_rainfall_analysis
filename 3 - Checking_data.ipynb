```python
precip = pd.read_excel(r"C:\Users\joaoa\Desktop\Doutoramento\Daily_rainfall_analysis\ERA5_Rainfall_PT_mm_Oct1981_Sep2023.xlsx")
precip
```

```python
max_diff = 0
n_errors = 0
prev_year = -1

for row in range(0, len(precip)): # range(0, len(precip))
    year = row//365 + 1981
    day = row - (row//365)*365 # Between 0 and 364
    
    if prev_year != year:
        df = pd.read_csv(r"D:\Precipitation\water_year\era5land_total_precipitation_" + str(year) + "_" + str(year+1) + "_(mm).csv", compression = "zip")
        print("Started water year " + str(year) + "_" + str(year+1))
    
    prev_year = year
    
    for col in range(1,1013):
        
        if round(precip.iloc[row, col], 2) == round(sum(df.iloc[(day*24):((day+1)*24), col]), 2):
            pass
        
        else:
            n_errors += 1
            diff = abs(precip.iloc[row, col] - sum(df.iloc[(day*24):((day+1)*24), col]))
            date_occ = "Year: " + str(year) + "_" + str(year+1) + " | Day: " + str(day) + " | " + precip.columns[col]

            if diff >= 0.01:
                    
                print(date_occ)
                print("Difference:", diff)
                print(" ")

                if diff > max_diff:
                    max_diff = diff
                    max_date_occ = date_occ

                    
print("Number of errors:", n_errors)
print(" ")
print("Maximum difference:", max_diff)
print("Date of occurence:", max_date_occ)
```
